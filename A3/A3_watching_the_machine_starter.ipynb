**What varied, and what held up?** *Summarize the divergence across runs and models, and how you verified each answer against ground truth.*

I asked each model to write Python code that summarizes a list of ten numbers: min, max, range, median, mode, Q1, Q3, IQR, and the sorted list. I chose this task because every value can be checked to determine if the model is correct.

I worked out the answers myself before comparing. The sorted list is 4.9, 5.6, 5.6, 6.4, 7.8, 7.8, 9.4, 14.3, 19.2, 21.2. The min is 4.9, the max is 21.2, and the range is 16.3. With ten values, the median is the average of the 5th and 6th values, which are both 7.8, so the median is 7.8. Two values appear twice (5.6 and 7.8), so there are two modes. For the quartiles I used the median-of-halves method. The lower half is 4.9, 5.6, 5.6, 6.4, 7.8, so Q1 = 5.6. The upper half is 7.8, 9.4, 14.3, 19.2, 21.2, so Q3 = 14.3. That makes the IQR 8.7. I also ran each model's code myself, so I wasn't relying on the expected output the model printed.

**For runs 1 to 3, using Claude**

All three Claude runs got the same numbers, and they all matched my hand calculation. When I ran the code from each run, the printed results matched what the model said they would be. All three used Python's built-in statistics module, all three used multimode so that both modes are reported, and all three warned that quartiles can be calculated in more than one way.

However there were some variations in the output:
- The code was different each time. Run 1 used statistics.median for everything. Runs 2 and 3 wrote their own median and quartiles helper functions, and run 3 split the upper half differently than run 2.
- The formatting was different. Run 1 rounded only the range and IQR. Run 2 rounded to 4 decimal places. Run 3 printed everything to 2 decimals (for example "7.80"). The order of the printed lines and the headers also changed.
- The side notes were different, and one of them was wrong. Every run mentioned that NumPy's default gives Q1 = 5.8 and Q3 = 13.075, and I confirmed that is correct. The main answer was right, but an extra detail the model added on its own was wrong, and it was stated just as confidently as everything else.
